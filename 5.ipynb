```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv.imread('new.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
plt.show()
```

```python
averaged_img = cv.blur(img, (50, 55))
plt.imshow(averaged_img)
plt.show()
```

```python
kernel = np.array([[4, 2, 4],
                   [2, 4, 2],
                   [4, 2, 4]], dtype=np.float32) / 16

weighted_img = cv.filter2D(img, -1, kernel)
plt.imshow(weighted_img)
plt.show()
```

```python
print(weighted_img)
```

```python
print(img)
```

```python
gaussian_img = cv.GaussianBlur(img, (31, 31), 0)
plt.imshow(gaussian_img)
plt.show()
```

```python
median_img = cv.medianBlur(img, 15)
plt.imshow(median_img)
plt.show()
```

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

sharpened_img = cv.filter2D(img, -1, kernel)

plt.imshow(sharpened_img)
plt.show()
```

```python
gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

laplacian_img = cv.Laplacian(gray_img, cv.CV_64F)

laplacian_img = cv.convertScaleAbs(laplacian_img)
plt.imshow(laplacian_img,cmap='gray')
plt.show()
```

```python

plt.figure(figsize=(15, 10))

images = [
    (img, "Original Image"),
    (averaged_img, "Averaging Filter"),
    (weighted_img, "Weighted Averaging"),
    (gaussian_img, "Gaussian Blur"),
    (median_img, "Median Blur"),
    (sharpened_img, "Laplacian Sharpening"),
    (gray_img, "Grayscale"),
    (laplacian_img, "Laplacian Operator")
]

for i, (image, title) in enumerate(images):
    plt.subplot(3, 3, i + 1)

    if len(image.shape) == 3:
        plt.imshow(image)
    else:
        plt.imshow(image, cmap='gray')

    plt.title(title)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
